```python
import pandas as pd
from PycalcAct.dataset import FromLegendFileCSV


trainingOptions = pd.read_csv("//Hmr_lymph/d/SebastienThis/CalciumPredictions/PycalcActivation/trainingOptions_round2.csv")

accuracy = trainingOptions["Accuracy"].values
customAcc = trainingOptions["Custom_Acc"].values
testingMetric = trainingOptions["Testing_Metric"].values
print(testingMetric)
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(20, 5))
ax = fig.add_subplot(131)
#scatter plot accuracy vs customAcc
plt.scatter(customAcc, 1/testingMetric, c=testingMetric, cmap='viridis', s=100)
plt.axis([0, 1, 0.75,1.25])
plt.xlabel("Custom Accuracy")
plt.ylabel("Testing Metric")

ax = fig.add_subplot(132)
plt.scatter(accuracy, 1/testingMetric, c=testingMetric, cmap='viridis', s=100)
plt.axis([0, 1, 0.75,1.25])
plt.xlabel("Accuracy")
plt.ylabel("Testing Metric")

ax = fig.add_subplot(133)
plt.scatter(accuracy, customAcc, c=testingMetric, cmap='viridis', s=100)
plt.axis([0, 1,0,1])
plt.xlabel("Accuracy")
plt.ylabel("Custom Accuracy")
```
